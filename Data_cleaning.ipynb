```python
import pandas as pd 
import numpy as np

df = pd.read_csv("Sample - Superstore.csv", encoding='latin-1')
```

```python
df.head()
```

```python
#  Removing any extra spaces around the column name, 
# lowering them so it can be in the easy format to work on sql later

df.columns = df.columns.str.strip().str.lower().str.replace(" ","_")

df.head(2)
```

```python
df.info()
```

```python
#  No nulls are available so thats good

df.isna().sum().sum()
```

```python
#  Changing to date format

df["order_date"] = pd.to_datetime(df["order_date"])

df["ship_date"] = pd.to_datetime(df["ship_date"])
```

```python
# Rounding the numbers into 2 decimals

df[["discount","profit","sales"]] = df[["discount","profit","sales"]].round(2)
```
